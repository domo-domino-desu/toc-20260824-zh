# Input and Output

Racket has input/output (I/O) procedures that will let you read from an *input port* or write to an *output port*. Ports can be associated with the console, files or strings.

##  Reading

Racket’s reader procedures take an optional input port argument. If the port is not specified, the default is the the current input port, usually the console.

Reading can be character-based, line-based or s-expression-based. Each time a read is performed, the port’s state changes so that the next read will read material following what was already read. If the port has no more material to be read then the reader procedure returns a specific datum called the end-of-file or *eof* object. This datum is the only value that satisfies the `eof‑object?` predicate.

The procedure `read‑char` reads the next character from the port. Read the next line with `read‑line`, and return it as a string (which does not include the final newline). Read the next s-expression with the procedure `read`.

##  Writing

Racket’s writer procedures take the object that is to be written and an optional output port argument. If the port is not specified then the default is the current output port, usually the console.

The writing can be character-based or s-expression-based.

The procedure `write‑char` writes the given character (without the `#\`) to the output port.

The procedures `write` and `display` both write the given s-expression to the port.  
The difference is that `write` tries to use a machine-readable format and `display` doesn’t. 
For instance, `write` uses double quotes for strings and the `#\\` syntax for characters, while `display` doesn’t.

The procedure `newline` starts a new line on the output port.


##  File ports

Scheme’s I/O procedures do not need a port argument if the port happens to be standard input or standard output. However, if you need these ports explicitly, the zero-argument procedures `current‑input‑port` and `current‑output‑port` furnish them. Thus, these two have the same behavior.

```
(display 9)
(display 9 (current-output-port))
```

A port is associated with a file by opening the file. The procedure `open‑input‑file` takes a filename argument and returns a new input port associated with it. Similarly, `open‑output‑file` takes a filename argument and returns a new output port associated with it. It is an error to open an input file that doesn’t exist, or to open an output file that already exists.

After you have performed I/O on a port, you should close it with `close‑input‑port` or `close‑output‑port`.

Assume the file *hello.txt* contains the single word "hello".
Then this will return the single character `#\h`.

```
(define i (open-input-file "hello.txt""))
(read-char i)
```

After those two, this command will return the string `ello`.

```
(define j (read i))
```

Here is an output example.
Assume that the file *greeting.txt* does not exist before the following programs are fed to the listener.

```
(define o (open-output-file "greeting.txt""))

(display "hello"" o)
(write-char #\space o)
(display 'world o)
(newline o)

(close-output-port o)
```

The file *greeting.txt* will now contain this single line.

```
hello world 
```

###  Automatic opening and closing of file ports

Racket supplies the procedures `call‑with‑input‑file` and `call‑with‑output‑file` that will take care of opening a port and closing it after you’re done with it.

For example, `call‑with‑input‑file` takes two arguments, a filename and a procedure. The procedure is applied to an input port opened on the file. When the procedure completes, its result is returned after ensuring that the port is closed.  Assume that *hello.txt* contains the string "hello".  Then this will return the list `(#\h #\e #\l)`.

```
(call-with-input-file "hello.txt""
  (lambda (i)
    (let* ((a (read-char i))
           (b (read-char i))
           (c (read-char i)))
      (list a b c))))
```  

## String ports

The procedure `open‑input‑string` associates a port with a given string. Reader procedures on this port will read off the string.  In this example, the string is "hello world" (not a file containing the string, just the string),
and will return `#\h`.

```
(define i (open-input-string "hello world""))
(read-char i)
```

Following that with this will return the string "ello".

```
(read i)
```

In turn, following with this will return "world".

```
(read i)
```

The procedure `open‑output‑string` creates an output port that will eventually be used to create a string.

```
(define o (open-output-string))

(write 'hello o)
(write-char #\, o)
(display " "" o)
(display "world"" o)
```

You can now use the procedure `get‑output‑string` to get the accumulated string in the string port `o`.
Thus, this

```
(get-output-string o)
```

returns "hello, world".  String ports need not be explicitly closed.


## Loading files

We have already seen the procedure `load` that loads files containing Racket code. Loading a file consists in evaluating in sequence every Racket form in the file. The pathname argument given to load is reckoned relative to the current working directory, which is normally the directory in which the Racket executable was called.

Files can load other files, and this is useful in a large program spanning many files. Unfortunately, unless full pathnames are used, the argument file of a `load` is dependent on Racket’s current directory.
