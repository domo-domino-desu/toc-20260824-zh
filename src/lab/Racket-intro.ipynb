# Data types

A *data type* is a collection of related values. One data type is the collection of integers, another is the collection of strings. These collections need not be disjoint, and they are often hierarchical. Racket has a rich set of data types: some are simple, that is, indivisible data types, such as integers or booleans. 

In some languages, programmers declare the data type of each variable before they use it.  But in Racket, you just starth using it, say by taking `x` to be `5`, and the system figures that since you used an integer, that is the type of the variable.  If you later use `x` as a string then the system switches the type.  For each data type, we will see how to test for that type.

One reason that the type matters is that different types allow different operations.  For each type we will describe the basic operations.

## Simple data types

These data types, such as integers or booleans, are indivisible.  That is, they are not made by combining other data types.

### Simple data type: Boolean

The booleans are `#t` for true and `#f` for false. 
These evaluate to themselves.

```racket
#t
```

Scheme has a procedure called `boolean?` that checks if its argument is boolean.

```racket
(boolean? #t)
```

```racket
(boolean? 42)
```

```racket
The procedure `not` negates its argument, considered as a boolean.
```

```racket
(not #f)
```

```racket
(not 42)
```

The last expression illustrates a Scheme convenience: in a context that requires a boolean, 
Scheme will treat any value other than `#f` as a true value.

The other natural Boolean operations are all available.  Here are the obvious ones.

```racket
(and #t #f)
```

```racket
(or #t #f)
```

```racket
#\c
```

```racket
#\newline
```

The character predicate is `char?`.

```racket
(char? #\c)
```

```racket
(char? 1)
```

```racket
(char? #\;)
```

(Note that a semicolon character datum does not trigger a comment.)
The character data type has its set of comparison predicates: `char=?`, `char<?`, `char<=?`, `char>?`, `char>=?`.

```racket
(char=? #\a #\a)
```

```racket
(char<? #\a #\b)
```

## Symbols

Symbols are used by Scheme programs as identifiers for variables.  But in contrast with the simple data types above that are self-evaluating, symbols evaluate to the value that the variable holds. That is, if `x` is a variable that is set to `7` and you ask to see `x`, then Racket will respond with `7`.  

Nevertheless, symbols are a simple data type, and symbols are legitimate values that Scheme can traffic in, along with characters, numbers, and the rest. To tell Racket that something is a symbol, not a variable, quote it.  (Here quoting means as in the sentence: You need an o to spell 'Ingenious'.)

```racket
(quote xyz)
```

Quoting is very common. 
There is a convenient abbreviation using the single-quote character.

```racket
'xyz
```

Scheme symbols are named by a sequence of characters. 
About the only limitation on a symbol’s name is that it shouldn’t be mistakable for some other data, such as characters, or booleans, or numbers, or compound data. Thus, `this‑is‑a‑symbol`, `i18n`, `<=>`, and `$!#*` are all symbols while `16`, `#t`, `"this‑is‑a‑string""`, `‑i` (which is a complex number), and `(barf)` (which is a list with one item) are not. 

The predicate for checking symbolness is `symbol?`

```racket
(symbol? 'xyz)
```

```racket
(symbol? 42)
```

Scheme symbols are normally case-insensitive. Thus the symbols `Calorie` and `calorie` are identical.

```racket
(eqv? 'Calorie 'calorie)
```

We can use a symbol such as `xyz` as a global variable by using the form define.

```racket
(define xyz 9)
```

This says the variable `xyz` holds the value `9`. If we feed `xyz` to the listener, the result will be the value held by `xyz`.

```racket
(define xyz 9)
xyz
```

```racket
(define xyz 9)
(set! xyz #\c)
xyz
```

(Note a difference between symbols and strings: symnbols are atomic, so you can't take them apart as you could by finding a substring, and you also cannot combine two together.)

## Strings

Strings are sequences of characters (not to be confused with symbols, which are simple data that have a sequence of characters as their name). You can specify strings by enclosing the constituent characters in double-quotes. Strings evaluate to themselves.

```racket
"Hello, World!"
```

The procedure `string` takes a bunch of characters and returns the string made from them.

```racket
(string #\h #\e #\l #\l #\o)
```

Let us now define a global variable `greeting`.

```racket
(define greeting "Hello; Hello!")
```

Note that a semicolon inside a string datum does not trigger a comment.

The characters in a given string can be individually accessed and modified. The procedure `string‑ref` takes a string and a (0-based) index, and returns the character at that index.

```racket
(define greeting "Hello; Hello!")
(string-ref greeting 0)
```

You can combine existing strings to make new strings.

```racket
(string-append "E "
               "Pluribus "
               "Unum")
```

You can make a string of a specified length and fill it with the desired characters later.

```racket
(define a-3-char-long-string (make-string 3))
a-3-char-long-string
```

(Explanation: the `\u` prefix means that the shown numbers are code points in Unicode.  So Racket is using the NULL character as a place-holder.)

The predicate for checking stringness is `string?`.

Strings obtained as a result of calls to `string`, `make‑string`, and `string‑append` are mutable. That is, they can change. 
Use the procedure `string‑set!` to replace the character at the given index (this index is 0-based).

```racket
(define hello (string #\H #\e #\l #\l #\o))
hello
```

```racket
(define hello (string #\H #\e #\l #\l #\o))
(string-set! hello 1 #\a)
hello
```

Strings can contain spaces and newline characters, by putting them between douple quotes.

```racket
(define hello-and-newline (string #\H #\e #\l #\l #\o #\newline #\t #\o #\d #\a #\y))
hello-and-newline  
```

Double quotation characters mark the string boundary, so to have a string that contains a double quote, you must *escape* that character: preceed it with a backslash character.

```racket
(string #\S #\a #\y #\space #\" #\A #\h #\" )
```

```racket
And what about backslash itself?
Use a second backslash. 
```

```racket
(define file-path (string #\c #\o #\n #\f #\\ #\i #\n #\i #\t))
file-path  
```

## Dotted pairs and lists


### Pairs

A *dotted pair* combines any two arbitrary values into an ordered couple. 
The first element is called the `car`, the second element is called the `cdr`, 
and the combining procedure is `cons`.
(These names involve the early history of the language.)

(cons 1 #t)
=> (1 . #t)


'(1 . #t) => (1 . #t)

(1 . #t)  -->ERROR!!!
The accessor procedures are car and cdr:

(define x (cons 1 #t))

(car x)
=> 1

(cdr x)
=> #t

The elements of a dotted pair can be replaced by the mutator procedures set‑car! and set‑cdr!:

(set-car! x 2)

(set-cdr! x #f)

x
=> (2 . #f)
Dotted pairs can contain other dotted pairs.

(define y (cons (cons 1 2) 3))

y
=> ((1 . 2) . 3)
The car of the car of this list is 1. The cdr of the car of this list is 2. I.e.,

(car (car y))
=> 1

(cdr (car y))
=> 2

Scheme provides procedure abbreviations for cascaded compositions of the car and cdr procedures. Thus, caar stands for “car of car of”, and cdar stands for “cdr of car of”, etc.

(caar y)
=> 1

(cdar y)
=> 2
c...r-style abbreviations for upto four cascades are guaranteed to exist. Thus, cadr, cdadr, and cdaddr are all valid. cdadadr might be pushing it.

When nested dotting occurs along the second element, Scheme uses a special notation to represent the resulting expression:

(cons 1 (cons 2 (cons 3 (cons 4 5))))
=> (1 2 3 4 . 5)
I.e., (1 2 3 4 . 5) is an abbreviation for (1 . (2 . (3 . (4 . 5)))). The last cdr of this expression is 5.


Scheme provides a further abbreviation if the last cdr is a special object called the empty list, which is represented by the expression (). The empty list is not considered self-evaluating, and so one should quote it when supplying it as a value in a program:

'() => ()
The abbreviation for a dotted pair of the form (1 . (2 . (3 . (4 . ())))) is

(1 2 3 4)
This special kind of nested dotted pair is called a list. This particular list is four elements long. It could have been created by saying

(cons 1 (cons 2 (cons 3 (cons 4 '()))))
but Scheme provides a procedure called list that makes list creation more convenient. list takes any number of arguments and returns the list containing them:

(list 1 2 3 4)
=> (1 2 3 4)
Indeed, if we know all the elements of a list, we can use quote to specify the list:

'(1 2 3 4)(cons 1 #t)
=> (1 2 3 4)
List elements can be accessed by index.

(define y (list 1 2 3 4))

(list-ref y 0) => 1
(list-ref y 3) => 4(1 . #t)

(list-tail y 1) => (2 3 4)
(list-tail y 3) => (4)
list‑tail returns the tail of the list starting from the given index.


The predicates pair?, list?, and null? check if their argument is a dotted pair, list, or the empty list, respectively:

(pair? '(1 . 2)) => #t
(pair? '(1 2))   => #t
(pair? '())      => #f
(list? '())      => #t
(null? '())      => #t
(list? '(1 2))   => #t
(list? '(1 . 2)) => #f
(null? '(1 2))   => #f
(null? '(1 . 2)) => #f

```racket
(cons 1 #t)
```

Dotted pairs are not self-evaluating and so to specify them directly as data (i.e., without producing them via a cons-call), you must explicitly quote them.

```racket
;; Just (1 . #t), without a quote, gives an error
'(1 . #t)
```

The accessor procedures are `car` and `cdr`.

```racket
(define x (cons 1 #t))
(car x)
```

```racket
(define x (cons 1 #t))
(cdr x)
```

Dotted pairs can contain other dotted pairs.

```racket
(define y (cons (cons 1 2) 3))
y
```

The `car` of the `car` of `y`. The `cdr` of the `car` is 2.

```racket
(define y (cons (cons 1 2) 3))
(car (car y))
```

```racket
(define y (cons (cons 1 2) 3))
(cdr (car y))
```

Racket provides procedure abbreviations for cascaded compositions of `car` and `cdr`. 
Thus, `caar` means “car of car of”, and `cdar` means “cdr of car of”, etc.

```racket
(define y (cons (cons 1 2) 3))
(caar y)
```

```racket
(define y (cons (cons 1 2) 3))
(cdar y)
```

The `c...r`-style abbreviations for up to four cascades are guaranteed to exist. Thus, `cadr`, `cdadr`, and `cdaddr` are all valid. 
But `cdadadr` might be pushing it.

### Lists

When nested dotting occurs along the second element, Racket uses a special notation to represent the resulting expression:

```racket
(cons 1 (cons 2 (cons 3 (cons 4 5))))
```

I.e., (1 2 3 4 . 5) is an abbreviation for (1 . (2 . (3 . (4 . 5)))). The last cdr of this expression is 5.

Racket provides a further abbreviation if the last cdr is a special object called the empty list, which is represented by the expression (). The empty list is not considered self-evaluating, and so one should quote it when supplying it as a value in a program.

```racket
'()
```

The abbreviation for a dotted pair of the form `(1 . (2 . (3 . (4 . ()))))` is
`(1 2 3 4)`
This special kind of nested dotted pair is called a *list*. 
This particular list is four elements long. 
It could have been created by saying
`(cons 1 (cons 2 (cons 3 (cons 4 '()))))`
but Scheme provides a procedure called `list` that makes list creation more convenient. 
The `list` procedure takes any number of arguments and returns the list containing them.

```racket
(list 1 2 3 4)
```

Indeed, if we know all the elements of a list, we can use quote to specify the list.

```racket
'(1 2 3 4)(cons 1 #t)
```

List elements can be accessed by index (as in other cases, the indexes are 0-based).

```racket
(define y (list 1 2 3 4))
(list-ref y 0)
```

```racket
(define y (list 1 2 3 4))
(list-ref y 3)
```

Use `list‑tail` to return the tail of the list starting from the given index.

```racket
(define y (list 1 2 3 4))
(list-tail y 1)
```

```racket
(define y (list 1 2 3 4))
(list-tail y 3)
```

The predicates `pair?`, `list?`, and `null?` check if their argument is a dotted pair, list, or the empty list, respectively.

```racket
(pair? '(1 . 2))
```

```racket
(pair? '(1 2))
```

```racket
(pair? '())
```

```racket
(list? '())
```

```racket
(null? '()) 
```

```racket
(list? '(1 2))
```

```racket
(list? '(1 . 2))  ;; does not have the right cdr
```

```racket
(null? '(1 2))
```

```racket
(null? '(1 . 2))
```

Dotted pairs, and lists, are immutable.  That is, once created, they cannot change.  This has many advantages, including speed.  For example, the procedure  `length` runs in constant time.  (You can get mutable pairs and lists by adding modules, but we will not do so.)

# Conversion among data types

Racket offers many procedures for converting among the data types. We already know how to convert between the character cases using char‑downcase and char‑upcase. Characters can be converted into integers using `char‑>integer`, and integers can be converted into characters using `integer‑>char`. (For us, the integer corresponding to a character is usually its ascii code.)

```racket
(char->integer #\d)
```

```racket
(integer->char 50)
```

You can convert a string into the corresponding list of characters.

```racket
(string->list "hello")
```

Other conversion procedures in the same vein are `list‑>string`, `vector‑>list`, and `list‑>vector`.

Numbers can be converted to strings:

```racket
(number->string 16)
```

You can also convert in the other direction, from strings to numbers. 
If the string corresponds to no number, Racket returns `#f`.

```racket
(string->number "16")
```

```racket
(string->number "Am I a number?")
```

```racket
The `string‑>number` procedure takes an optional second argument, the radix.
Below, Racket gives 14 because in base 8, the string "16" represents the number we know as 14.
```

```racket
(string->number "16" 8)
```

You can also interconvert between symbols and strings.

```racket
(symbol->string 'symbol)
```

```racket
(string->symbol "string")
```

# Other data types

Racket contains some other data types. One is the procedure. We have already seen many procedures, e.g., `display`, `+`, and `cons`. In reality, these are variables holding the procedure values, which are themselves not visible as are numbers or characters:

```racket
cons
```

The procedures we have seen thus far are primitive procedures, with standard global variables holding them. Users can create additional procedure values.

Yet another data type is the `port`. A `port` is the conduit through which Racket performs input and output. They are usually associated with files and consoles.

In our "Hello, World!" program, we used the procedure `display` to write a string to the console. This procedure can take two arguments, one the value to be displayed, and the other the output port that it should be displayed on.

In our program, display’s second argument was implicit. The default output port used is the standard output port. We can get the current standard output port via the procedure-call `(current‑output‑port)`. We could have been more explicit and written this.


```racket
(display "Hello, World!"" (current-output-port))
```

# S-expressions

All the data types discussed here can be lumped together into a single all-encompassing data type called the *s-expression* (s for symbolic). Thus `42`, `#\c`, `(1 . 2)`, `#(a b c)`, `"Hello"`, `(quote xyz)`, `(string‑>number "16")`, and indeed `(begin (display "Hello, World!"") (newline))` are all s-expressions.
