# Data types

A *data type* is a collection of related values. One data type is the collection of integers, another is the collection of strings. These collections need not be disjoint, and they are often hierarchical. Racket has a rich set of data types: some are simple, that is, indivisible data types, such as integers or booleans. 

In some languages, programmers declare the data type of each variable before they use it.  But in Racket, you just starth using it, say by taking `x` to be `5`, and the system figures that since you used an integer, that is the type of the variable.  If you later use `x` as a string then the system switches the type.  For each data type, we will see how to test for that type.

One reason that the type matters is that different types allow different operations.  For each type we will describe the basic operations.

## Simple data types

These data types, such as integers or booleans, are indivisible.  That is, they are not made by combining other data types.

### Simple data type: Boolean

The booleans are `#t` for true and `#f` for false. 
These evaluate to themselves.

```racket
#t
```

Scheme has a procedure called `boolean?` that checks if its argument is boolean.

```racket
(boolean? #t)
```

```racket
(boolean? 42)
```

```racket
The procedure `not` negates its argument, considered as a boolean.
```

```racket
(not #f)
```

```racket
(not 42)
```

The last expression illustrates a Scheme convenience: in a context that requires a boolean, 
Scheme will treat any value other than `#f` as a true value.

The other natural Boolean operations are all available.  Here are the obvious ones.

```racket
(and #t #f)
```

```racket
(or #t #f)
```

### Simple data type: Numbers

By numbers we mean integers, or fractions such as 22/7, or floating point numbers such as 3.14159, or complex numers such as 3+4i. However, we will mostly stick to integers.  

As with Booleans, numbers evaluate to themselves.  Thus, evaluating the 42 here returns 42. 

```racket
42
```

The data type of numbers is hierarchical: an integer is a rational, a rational is a real, and a real is a complex number, and all of them are numbers. 

Racket has predicates functions for testing the various flavors of number.

```racket
(number? 42) 
```

```racket
(number? #t)
```

```racket
(complex? 2+3i)
```

```racket
(real? 2+3i)
```

```racket
(real? 22/7)
```

```racket
real? 3.1416) 
```

```racket
(rational? 22/7)
```

```racket
(integer? 22/7)
```

```racket
(integer? 42)
```

For integers, you can use bases other than 10.  Get binary by prefixing the numeral with `#b` so that `#b1100` is the number twelve. The octal prefix is `#o` and the hex prefix is `#x`.

```racket
#b101010
```

You can test numbers for equality using the general-purpose equality predicate `eqv?`.

```racket
(eqv? 42 42)
```

```racket
(eqv? 42 #f)
```

```racket
(eqv? 42 42.0)
```

```racket
(= 42 42)
```

```racket
(= 42 #f)
```

```racket
(= 42 42.0)
```

Other available number comparisons are `<`, `<=`, `>`, and `>=`.

```racket
(< 3 2) 
```

## Characters

To represent character data use a `#\` prefix, as in `#\a` or `#\b` for `a` or `b`. Some characters have more descriptive names: `#\newline` and `#\tab`, for instance. For a space character, write `#\space` (or `#\ `).

```racket
#\c
```

```racket
#\newline
```

The character predicate is `char?`.

```racket
(char? #\c)
```

```racket
(char? 1)
```

```racket
(char? #\;)
```

(Note that a semicolon character datum does not trigger a comment.)
The character data type has its set of comparison predicates: `char=?`, `char<?`, `char<=?`, `char>?`, `char>=?`.

```racket
(char=? #\a #\a)
```

```racket
(char<? #\a #\b)
```

```racket
(char>=? #\a #\b)
```

Make the comparisons case-insensitive with  `char‑ci` instead of `char`.

```racket
(char-ci=? #\a #\A)
```

```racket
(char-ci<? #\a #\B)
```

Convert cases with `char‑downcase` and `char‑upcase`.

```racket
(char-downcase #\A)
```

```racket
(char-upcase #\a) 
```

## Symbols

Symbols are used by Scheme programs as identifiers for variables.  But in contrast with the simple data types above that are self-evaluating, symbols evaluate to the value that the variable holds. That is, if `x` is a variable that is set to `7` and you ask to see `x`, then Racket will respond with `7`.  

Nevertheless, symbols are a simple data type, and symbols are legitimate values that Scheme can traffic in, along with characters, numbers, and the rest. To tell Racket that something is a symbol, not a variable, quote it.  (Here quoting means as in the sentence: You need an o to spell 'Ingenious'.)

```racket
(quote xyz)
```

Quoting is very common. 
There is a convenient abbreviation using the single-quote character.

```racket
'xyz
```

Scheme symbols are named by a sequence of characters. 
About the only limitation on a symbol’s name is that it shouldn’t be mistakable for some other data, such as characters, or booleans, or numbers, or compound data. Thus, `this‑is‑a‑symbol`, `i18n`, `<=>`, and `$!#*` are all symbols while `16`, `#t`, `"this‑is‑a‑string""`, `‑i` (which is a complex number), and `(barf)` (which is a list with one item) are not. 

The predicate for checking symbolness is `symbol?`

```racket
(symbol? 'xyz)
```

```racket
(symbol? 42)
```

Scheme symbols are normally case-insensitive. Thus the symbols `Calorie` and `calorie` are identical.

```racket
(eqv? 'Calorie 'calorie)
```

We can use a symbol such as `xyz` as a global variable by using the form define.

```racket
(define xyz 9)
```

This says the variable `xyz` holds the value `9`. If we feed `xyz` to the listener, the result will be the value held by `xyz`.

```racket
(define xyz 9)
xyz
```

```racket
(define xyz 9)
(set! xyz #\c)
xyz
```

(Note a difference between symbols and strings: symnbols are atomic, so you can't take them apart as you could by finding a substring, and you also cannot combine two together.)

## Compound data types

In contrast to the simple data types above, compound data types are made by combining other data types, such as that strings are sequences of characters.

## Strings

Strings are sequences of characters (not to be confused with symbols, which are simple data that have a sequence of characters as their name). You can specify strings by enclosing the constituent characters in double-quotes. Strings evaluate to themselves.

```racket
"Hello, World!"
```

The procedure string takes a bunch of characters and returns the string made from them.

```racket
(string #\h #\e #\l #\l #\o)
```

Let us now define a global variable `greeting`.

```racket
(define greeting "Hello; Hello!")
```

Note that a semicolon inside a string datum does not trigger a comment.

The characters in a given string can be individually accessed and modified. The procedure `string‑ref` takes a string and a (0-based) index, and returns the character at that index.

```racket
(define greeting "Hello; Hello!")
(string-ref greeting 0)
```

You can combine existing strings to make new strings.

```racket
(string-append "E "
               "Pluribus "
               "Unum")
```

You can make a string of a specified length and fill it with the desired characters later.

```racket
(define a-3-char-long-string (make-string 3))
a-3-char-long-string
```

(Explanation: the `\u` prefix means that the shown numbers are code points in Unicode.  So Racket is using the NULL character as a place-holder.)

The predicate for checking stringness is `string?`.

Strings obtained as a result of calls to `string`, `make‑string`, and `string‑append` are mutable. That is, they can change. 
Use the procedure `string‑set!` to replace the character at the given index (this index is 0-based).

```racket
(define hello (string #\H #\e #\l #\l #\o))
hello
```

```racket
(define hello (string #\H #\e #\l #\l #\o))
(string-set! hello 1 #\a)
hello
```

Strings can contain spaces and newline characters, by putting them between douple quotes.

```racket
(define hello-and-newline (string #\H #\e #\l #\l #\o #\newline #\t #\o #\d #\a #\y))
hello-and-newline  
```

Double quotation characters mark the string boundary, so to have a string that contains a double quote, you must *escape* that character: preceed it with a backslash character.

```racket
(string #\S #\a #\y #\space #\" #\A #\h #\" )
```

```racket
And what about backslash itself?
Use a second backslash. 
```

```racket
(define file-path (string #\c #\o #\n #\f #\\ #\i #\n #\i #\t))
file-path  
```
